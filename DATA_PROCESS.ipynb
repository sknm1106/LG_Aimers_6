````python
df['총 시술 횟수'] = df['총 시술 횟수'].str.extract('(\d+)').astype(float)
df['IVF 시술 횟수'] = df['IVF 시술 횟수'].str.extract('(\d+)').astype(float)
df['IVF 임신 횟수'] = df['IVF 임신 횟수'].str.extract('(\d+)').astype(float)
df['DI 시술 횟수'] = df['DI 시술 횟수'].str.extract('(\d+)').astype(float)
df['DI 임신 횟수'] = df['DI 임신 횟수'].str.extract('(\d+)').astype(float)
df['총 임신 횟수'] = df['총 임신 횟수'].str.extract('(\d+)').astype(float)
df['클리닉 내 총 시술 횟수'] = df['클리닉 내 총 시술 횟수'].str.extract('(\d+)').astype(float)
df['총 출산 횟수'] = df['총 출산 횟수'].str.extract('(\d+)').astype(float)
df['DI 출산 횟수'] = df['DI 출산 횟수'].str.extract('(\d+)').astype(float)
df['IVF 출산 횟수'] = df['IVF 출산 횟수'].str.extract('(\d+)').astype(float)
````

````python
def extract_age(text):
    numbers = list(map(int, re.findall(r'\d+', text)))  # 모든 숫자 추출
    if numbers:
        return np.mean(numbers)  # 평균 반환
    return 0
````

````python
df.loc[(df['기증 배아 사용 여부'] == 1.0), '난자 출처'] = '기증 제공'
df.loc[(df['기증 배아 사용 여부'] == 1.0), '정자 출처'] = '기증 제공'

df.loc[df['난자 출처'] == '알 수 없음', '난자 출처'] = '본인 제공'

df.loc[
    (df['정자 출처'] == '미할당') &
    (df['정자 기증자 나이'] != 0) &
    (df['파트너 정자와 혼합된 난자 수'] == 0),
    '정자 출처'
] = '기증 제공'
df.loc[
    (df['정자 출처'] == '미할당') &
    (df['정자 기증자 나이'] != 0) &
    (df['파트너 정자와 혼합된 난자 수'] == 1),
    '정자 출처'
] = '배우자 및 기증 제공'
df.loc[
    (df['정자 출처'] == '미할당') &
    (df['정자 기증자 나이'] == 0) &
    (df['파트너 정자와 혼합된 난자 수'] == 1),
    '정자 출처'
] = '배우자 제공'
df.loc[
    (df['정자 출처'] == '미할당') &
    (df['정자 기증자 나이'] != 0) &
    (df['파트너 정자와 혼합된 난자 수'] == 0),
    '정자 출처'
] = '기증 제공'
df.loc[df['정자 출처'] == '미할당', '정자 출처'] = '배우자 제공'

# One-Hot Encoding
one_hot_features=['난자 출처','정자 출처']
df = pd.get_dummies(df, columns=one_hot_features)
````

````python
def process_test_data(df):
  df.drop(columns=['시술 시기 코드', '임신 시도 또는 마지막 임신 경과 연수' ,'불임 원인 - 여성 요인', '불임 원인 - 정자 면역학적 요인', '불임 원인 - 자궁경부 문제'], inplace=True)
  df['총 시술 횟수'] = df['총 시술 횟수'].str.extract('(\d+)').astype(float)
  df['IVF 시술 횟수'] = df['IVF 시술 횟수'].str.extract('(\d+)').astype(float)
  df['IVF 임신 횟수'] = df['IVF 임신 횟수'].str.extract('(\d+)').astype(float)
  df['DI 시술 횟수'] = df['DI 시술 횟수'].str.extract('(\d+)').astype(float)
  df['DI 임신 횟수'] = df['DI 임신 횟수'].str.extract('(\d+)').astype(float)
  df['총 임신 횟수'] = df['총 임신 횟수'].str.extract('(\d+)').astype(float)
  df['클리닉 내 총 시술 횟수'] = df['클리닉 내 총 시술 횟수'].str.extract('(\d+)').astype(float)
  df['총 출산 횟수'] = df['총 출산 횟수'].str.extract('(\d+)').astype(float)
  df['DI 출산 횟수'] = df['DI 출산 횟수'].str.extract('(\d+)').astype(float)
  df['IVF 출산 횟수'] = df['IVF 출산 횟수'].str.extract('(\d+)').astype(float)

  df['난자 기증자 나이'] = df['난자 기증자 나이'].apply(extract_age)
  df['정자 기증자 나이'] = df['정자 기증자 나이'].apply(extract_age)

  df['시술 당시 나이'] = df['시술 당시 나이'].apply(extract_age)
  df.loc[df['시술 당시 나이'] == '알 수 없음', '시술 당시 나이'] = 50

  df['특정 시술 유형'] = df['특정 시술 유형'].fillna('Unknown')
  df['세부 조합 시행 여부'] = df['특정 시술 유형'].str.contains(':').astype(int)
  df['복합적인 시술 시행 여부'] = df['특정 시술 유형'].str.contains('/').astype(int)
  df = df.drop(columns = ['특정 시술 유형','시술 유형'])

  df['현재 시술용 배아 생성'] = df["배아 생성 주요 이유"].apply(lambda x: 1 if "현재 시술용" in str(x) else 0)
  df['배아/난자 저장 또는 기증 여부'] = df["배아 생성 주요 이유"].apply(
    lambda x: 1 if any(keyword in str(x) for keyword in ["배아 저장용", "난자 저장용", "기증용"]) else 0
  )
  df = df.drop(columns = ['배아 생성 주요 이유'])

  filtered_df = df[
    (df['착상 전 유전 진단 사용 여부'] == 0) &
    (df['착상 전 유전 검사 사용 여부'] == 1) &
    (df['불임 원인 - 정자 농도'] == 0) &
    (df['불임 원인 - 정자 운동성'] == 0) &
    (df['불임 원인 - 정자 형태'] == 0) &
    (df['PGD 시술 여부'].isnull()) &
    ((df['난자 채취 경과일'] == 0) | (df['난자 채취 경과일'].isnull())) &
    (df['현재 시술용 배아 생성'] == 1) &
    (df['배아/난자 저장 또는 기증 여부'] == 0)
  ]
  df.loc[filtered_df.index, 'PGS 시술 여부'] = 1.0

  filtered_df = df[
    (df['착상 전 유전 진단 사용 여부'] == 1) &
    (df['착상 전 유전 검사 사용 여부'] == 0) &
    (df['불임 원인 - 정자 농도'] == 0) &
    (df['불임 원인 - 정자 운동성'] == 0) &
    (df['불임 원인 - 정자 형태'] == 0) &
    (df['PGS 시술 여부'].isnull())
  ]
  df.loc[filtered_df.index, 'PGD 시술 여부'] = 1.0

  df["남성 요인 불임 여부"] = df[["남성 주 불임 원인", "남성 부 불임 원인", '불임 원인 - 남성 요인', '불임 원인 - 정자 농도', '불임 원인 - 정자 운동성', '불임 원인 - 정자 형태']].max(axis=1)
  df["여성 요인 불임 여부"] = df[["여성 주 불임 원인", "여성 부 불임 원인",'불임 원인 - 난관 질환', '불임 원인 - 배란 장애', '불임 원인 - 자궁내막증']].max(axis=1)
  df["불명확 원인 불임 여부"] = df["불명확 불임 원인"]
  df = df.drop(columns = ['남성 주 불임 원인', '남성 부 불임 원인', '여성 주 불임 원인', '여성 부 불임 원인', '불명확 불임 원인',
                        '불임 원인 - 남성 요인', '불임 원인 - 정자 농도', '불임 원인 - 정자 운동성', '불임 원인 - 정자 형태',
                        '불임 원인 - 난관 질환', '불임 원인 - 배란 장애', '불임 원인 - 자궁내막증'])

  df = df.drop(columns=['착상 전 유전 검사 사용 여부'])
  df.loc[(df['해동된 배아 수'] == 0.0) & (df['동결 배아 사용 여부'].isna()), '동결 배아 사용 여부'] = 0.0
  df.loc[(df['총 생성 배아 수'] == 0.0) & (df['신선 배아 사용 여부'].isna()), '신선 배아 사용 여부'] = 0.0

  df.loc[(df['기증 배아 사용 여부'] == 1.0), '난자 출처'] = '기증 제공'
  df.loc[(df['기증 배아 사용 여부'] == 1.0), '정자 출처'] = '기증 제공'

  df.loc[df['난자 출처'] == '알 수 없음', '난자 출처'] = '본인 제공'

  df.loc[
    (df['정자 출처'] == '미할당') &
    (df['정자 기증자 나이'] != 0) &
    (df['파트너 정자와 혼합된 난자 수'] == 0),
    '정자 출처'
  ] = '기증 제공'
  df.loc[
    (df['정자 출처'] == '미할당') &
    (df['정자 기증자 나이'] != 0) &
    (df['파트너 정자와 혼합된 난자 수'] == 1),
    '정자 출처'
  ] = '배우자 및 기증 제공'
  df.loc[
    (df['정자 출처'] == '미할당') &
    (df['정자 기증자 나이'] == 0) &
    (df['파트너 정자와 혼합된 난자 수'] == 1),
    '정자 출처'
  ] = '배우자 제공'
  df.loc[
    (df['정자 출처'] == '미할당') &
    (df['정자 기증자 나이'] != 0) &
    (df['파트너 정자와 혼합된 난자 수'] == 0),
    '정자 출처'
  ] = '기증 제공'
  df.loc[df['정자 출처'] == '미할당', '정자 출처'] = '배우자 제공'
  # One-Hot Encoding
  one_hot_features=['난자 출처','정자 출처']
  df = pd.get_dummies(df, columns=one_hot_features)
  df.loc[
    ((df['신선 배아 사용 여부'] == 1) | (df['동결 배아 사용 여부'] == 1)) & (df['대리모 여부'].isnull()),
    '대리모 여부'
  ] = 1

  df = df.drop(columns = ['난자 채취 경과일'])
  df['배란 유도 유형'] = df['배란 유도 유형'].apply(lambda x: 0 if x == "알 수 없음" else 1)

  df=df.fillna(0)
  return df
````
